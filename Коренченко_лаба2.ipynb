Оценка пустых ячеек в датасете

````python
def null_count(data):
  rows =[]
  for col in data.columns:
    column = data[col]
    null = column.isna().sum()
    if null > 0:
      null_count = null
    elif null == 0:
      null_count = 0
    column_count = len(column)
    rows.append(
        {'Столбец': col,
        'Кол-во NULL': null_count,
        'Кол-во строк': column_count}
    )
  return pd.DataFrame(rows)

def null_75(data):
  null = data['Кол-во NULL']
  full = data['Кол-во строк']
  null_pr = null*100/full.where(full>0)
  return (null>75).sum().item()

null_VS_full = null_count(data)
print(null_VS_full)
print(f'Количество строк, в которых пустыми являются более 75% ячеек: {null_75(null_VS_full)}')

fig, ax = plt.subplots(figsize=(12, 5))
msno.bar(
    data,
    ax = ax,
    color='pink',
    fontsize=10,
    labels=True
)
ax.set_title('Заполненность столбцов')
plt.show()
````

Ознакомиться с теоретическим материалом по работе с регулярными выражениями можно [по ссылке](https://colab.research.google.com/drive/1eFHdmt9OROLEDJRNc2wNmz0gpE1O3sKU?usp=sharing).

````python
patterns = {
    'город-код штата-индекс': r'[A-Za-z][A-Za-z .\'-]*,\s*[A-Z]{2}\s+\d{5}',
    'код штата-индекс': r'[A-Z]{2}\s+\d{5}',
    'город-код штата': r'[A-Za-z][A-Za-z .\'-]*,\s*[A-Z]{2}',
    'город-индекс': r'[A-Za-z][A-Za-z .\'-]*,?\s*\d{5}',
    'индекс': r'\d{5}',
    'код штата': r'[A-Z]{2}',
    'город': r'[A-Za-z][A-Za-z .\'-]*',
}

location_values = data['location'].fillna('').astype(str)
d_format = pd.Series('не определен', index=location_values.index)

for format_name, pattern in patterns.items():
    is_not_detected_yet = (d_format == 'не определен')
    matches_current_pattern = location_values.str.fullmatch(pattern, na=False)
    rows_to_label = is_not_detected_yet & matches_current_pattern
    d_format.loc[rows_to_label] = format_name

format_counts = d_format.value_counts()
print(format_counts)

format_counts.plot(kind='bar', figsize=(8, 4))
plt.title('Форматы столбца location')
plt.xlabel('Формат')
plt.ylabel('Количество строк')
plt.show()
````

В большей части строк удалось извлечь данные из ячейки столбца `location`, но остались и такие, где это не получилось.

1. Вынесите в отдельный датасет записи, в которых не удалось определить формат записи информации о местоположении.
2. Данные о местоположении разделите на 3 столбца: город, код штата, почтовый индекс.

````python
unrecognized_location_data = data.loc[d_format == 'не определен'].copy()
new_data = data.loc[d_format != 'не определен'].copy()

city  = pd.Series(np.nan, index=location_values.index, dtype=object)
state = pd.Series(np.nan, index=location_values.index, dtype=object)
zipc  = pd.Series(np.nan, index=location_values.index, dtype=object)

mask_city = d_format.isin([
    'город-код штата-индекс',
    'город-код штата',
    'город-индекс',
    'город',
])
city[mask_city] = location_values[mask_city].str.extract(
    r'^([A-Za-z][A-Za-z .\'-]*?)\s*(?:,|$|\s\d)',
    expand=False
)

mask_state = d_format.isin([
    'город-код штата-индекс',
    'код штата-индекс',
    'город-код штата',
    'код штата',
])
state[mask_state] = location_values[mask_state].str.extract(
    r'\b([A-Z]{2})\b',
    expand=False
)

mask_zip = d_format.isin([
    'город-код штата-индекс',
    'код штата-индекс',
    'город-индекс',
    'индекс',
])
zipc[mask_zip] = location_values[mask_zip].str.extract(
    r'(\d{5})',
    expand=False
)

new_data['город'] = city
new_data['код штата'] = state
new_data['индекс'] = zipc
new_data = new_data.drop(columns=['location'])

print(new_data.sample(5))
````
